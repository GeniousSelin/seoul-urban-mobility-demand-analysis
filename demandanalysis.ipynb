**Exercise 9**

Interquartile range:

```python
q1 = df['Temperature(캜)'].quantile(0.25)
q3 = df['Temperature(캜)'].quantile(0.75)
iqr = q3 - q1
lower = q3 - 1.5 * iqr
upper = q3 + 1.5 * iqr
v = df[(df['Temperature(캜)'] < lower) | (df['Temperature(캜)'] > upper)]
v.shape[0]
```

```python
q1 = df['Dew point temperature(캜)'].quantile(0.25)
q3 = df['Dew point temperature(캜)'].quantile(0.75)
iqr = q3 - q1
lower = q3 - 1.5 * iqr
upper = q3 + 1.5 * iqr
v = df[(df['Dew point temperature(캜)'] < lower) | (df['Dew point temperature(캜)'] > upper)]
v.shape[0]
```

Standard deviation:

```python
avg = df['Temperature(캜)'].mean()
s = df['Temperature(캜)'].std()
lower = avg - 1.5 * s
upper = avg + 1.5 * s
v = df[(df['Temperature(캜)'] < lower) | (df['Temperature(캜)'] > upper)]
v.shape[0]
```

```python
avg = df['Dew point temperature(캜)'].mean()
s = df['Dew point temperature(캜)'].std()
lower = avg - 1.5 * s
upper = avg + 1.5 * s
v = df[(df['Dew point temperature(캜)'] < lower) | (df['Dew point temperature(캜)'] > upper)]
v.shape[0]
```

Based on the results, the number of outliers differs between the two detection methods. Using the _Interquartile range method (IQR)_, there were **555** outliers detected for Temperature(캜) and **985** outliers for Dew point temperature(캜). In contrast, when using the _Standard deviation method_, there were **1125** outliers for Temperature(캜) and **1057** outliers for Dew point temperature(캜). This shows that the standard deviation method identified noticeably more outliers, especially for temperature. The difference occurs because the IQR method is based on quartiles and is less sensitive to extreme values, while the standard deviation method depends on the mean and standard deviation, which can be influenced by the presence of outliers or non-normal data distribution. Therefore, the IQR method provides a more robust estimate of outliers for skewed or non-normal data, whereas the standard deviation method is more appropriate when the data are approximately normally distributed.

**Exercise 2.1**

```python
from sklearn.preprocessing import LabelEncoder

df1['Temp_Category_coded'] = LabelEncoder().fit_transform(df1['Temp_Category'])
df1['Time_coded'] =  LabelEncoder().fit_transform(df1['Time'])
df1['Rainfall_Level_coded'] = LabelEncoder().fit_transform(df1['Rainfall_Level'])
df1['Seasons_coded'] = LabelEncoder().fit_transform(df1['Seasons'])
df1['Holiday_coded'] = LabelEncoder().fit_transform(df1['Holiday'])
df1['Functioning Day_coded'] = LabelEncoder().fit_transform(df1['Functioning Day'])
df1
```
